#### PySpark code executed in the Databricks Cluster

```python
companies_df = spark.read.table("demo.silver.companies") 
summary_df = companies_df.groupBy("country").count()
summary_py = summary_df.collect()   
```

#### Python code executed in the local machine

```python
total_companies = sum(row['count'] for row in summary_py)
print("Total companies across all countries:", total_companies)
```
