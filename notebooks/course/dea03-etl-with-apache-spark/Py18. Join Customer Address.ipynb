## Join Customer and Address
Join customer data with address data to create a customer_address table which contains the address of each customer on the same record

## Learning notes — before source operation

````python
df_customers = spark.table('gizmobox.silver.py_customers')
display(df_customers)
````

## Learning notes — before source operation

**profiling**: Read the selected columns and output grain before interpreting the saved result. Counts, nulls and samples describe the archived input state, not a newly executed validation.

````python
df_addresses = spark.table('gizmobox.silver.py_addresses')
display(df_addresses)
````

## Learning notes — before source operation

**join**: Check join keys and cardinality on both inputs; unmatched keys and duplicate keys affect nulls and row counts in the joined output.

**Unexecuted teaching example (illustrative, not an archived or generated result):** Two left rows and three right rows with the same matching key yield six joined rows.

**profiling**: Read the selected columns and output grain before interpreting the saved result. Counts, nulls and samples describe the archived input state, not a newly executed validation.

````python
df_customer_address = (
    df_customers.join(df_addresses, "customer_id", "inner")
    .select(
        "customer_id",
        "customer_name",
        "email",
        "date_of_birth",
        "member_since",
        "telephone",
        "shipping_address_line_1",
        "shipping_city",
        "shipping_state",
        "shipping_postcode",
        "billing_address_line_1",
        "billing_city",
        "billing_state",
        "billing_postcode"        
    )
)
display(df_customer_address)
````

````python
df_customer_address.writeTo("gizmobox.gold.py_customer_address").createOrReplace()
````
